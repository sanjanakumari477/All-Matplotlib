```python
import matplotlib.pyplot as plt

# List of product names
products = ["Laptop", "Smartphone", "Headphones", "Smartwatch", "Tablet",
            "Keyboard", "Mouse", "Monitor", "Printer", "External Hard Drive"]

# List of number of sales (same order as products)
sales = [120, 250, 180, 95, 130, 75, 110, 90, 60, 40]

# Correct bar plot
plt.bar(products, sales, color='purple', label='Sales 2025')

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales Report 2025")
plt.legend(loc='upper right', fontsize=12)
plt.xticks(rotation=45)  # rotate x-axis labels for better view
plt.tight_layout()       # adjust layout
plt.show()
```
