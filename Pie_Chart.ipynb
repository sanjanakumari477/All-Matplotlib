```python
import matplotlib.pyplot as plt

# Data for pie chart
brands = ["Samsung", "Apple", "Xiaomi", "Oppo", "Vivo", "OnePlus"]
market_share = [28, 24, 18, 12, 10, 8]

# Create pie chart
plt.pie(market_share, labels=brands, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title("Smartphone Market Share 2025")
plt.show()
```
